```python
score = 0
print("Welcome to the Quiz Game😄")
print()
answer = input("1.What's the capital of Japan?")
if answer.lower() == "tokyo":
    print("Your answer is correct")
    score = score + 1
else:
    print("Wrong Answer!The correct one is Tokyo")

answer = input("2.What's 12 * 5?")
if answer == "60":
    print("Your answer is correct")
    score = score + 1
else:
    print("Wrong Answer!The correct one is 60")

answer = input("3.What's the currency of SouthKorea?")
if answer.lower() == "won":
    print("Your answer is correct")
    score = score + 1
else:
    print("Wrong Answer!The correct one is Won")

print()
print("Quiz finished!")
print("Your Score is:",score,"/3")
```
